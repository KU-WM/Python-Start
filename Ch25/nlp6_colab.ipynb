## Seq2Seq (Sequence-to-Sequence)

> 입력 시퀀스를 받아 출력 시퀀스로 변환하는 인코더-디코더 구조의 신경망 모델 <br>
> 입력 시퀀스와 출력 시퀀스의 길이가 다른 문제에 적합한 모델 <br>
> 예) 문장 번역, 문장 요약, 질의 응답

### ✅ Seq2Seq 모델 특징

#### ① 시퀀스 길이가 달라져도 처리 가능
- 기존의 고정 길이 입력/출력 구조에서는 불가능했던 자연어 처리 문제 해결 가능

#### ② 인코더-디코더 분리 구조
- 인코더, 디코더는 RNN 계열의 순환 신경망 또는 Transformer 계열 모델로 구성
- 인코더는 입력 시퀀스를 이해하고 요약 (context vector 생성)
- 디코더는 context vector를 기반으로 출력 시퀀스 생성

#### ③ Teaching Forcing
- 디코더가 이전 단계의 실제 정답 토큰을 다음 단계의 입력으로 사용하는 학습 기법
- 학습 중 디코더가 정답 토큰을 다음 입력으로 받게 하여 더 빠르고 안정적인 학습을 도와주는 기법
- 실제로 구현할때는 무조건 정답을 주지 않고, 일정 확률로 정답을 넣어주거나 예측값을 넣어줌으로써 모델이 빠르게 학습을 하되 실전에서도 잘 예측할 수 있도록 균형잡힌 학습 전략을 사용함

#### ④ Auto-Regressive
- 이전 시점까지의 출력을 바탕으로 다음 시점의 출력을 예측하는 방식
- 디코더가 한 단어씩 순차적으로 생성하기 때문에 병렬화가 어려움

#### ⑤ 정보 압축의 문제
- 모든 정보를 하나의 context vector(h_n, c_n)에 요약하는 구조의 한계가 있음
- 긴 문장일 수록 정보 소실이 발생할 가능성이 큼

#### ⑥ Exposure Bias (예측 누적 오류 문제)
- 학습 시에 정답 시퀀스를 디코더의 입력으로 넣어 주지만(Teaching Forcing), <br>
 추론 시에는 모델의 예측 값을 디코더의 입력으로 넣어야 하는 구조로 인해 발생하는 예측 오차 누적 현상
- 예측 값이 시퀀스 데이터이기 때문에 시퀀스의 앞부분의 예측이 잘못 되면 뒤에 오는 예측 결과에 영향을 받음

#### ⑦ 순차 처리
- 인코더와 디코더 모두 단어 하나씩 순차적으로 처리.
- 병렬화가 불가능
- 훈련/추론 속도가 느림

### LSTM vs Seq2Seq

| 항목 | LSTM | Seq2Seq |
|------|------|---------|
| **목적** | 시계열/시퀀스 데이터 처리 | 입력 시퀀스를 출력 시퀀스로 변환 |
| **구조** | 하나의 RNN(LSTM) 구조 | 인코더-디코더 두 개의 LSTM |
| **입출력 길이** | 대체로 동일 | 서로 달라도 가능 |
| **정보 흐름** | t-1 시점 → t 시점 | 인코더 전체 → context vector → 디코더 |
| **학습 기법** | 일반적인 시퀀스 학습 | Teacher Forcing 사용 |
| **사용 예시** | 감정 분류, 시계열 예측 | 기계 번역, 텍스트 요약, 질의응답 등 |

## Seq2Seq 구조 시각화 및 수식

<center> Encoder </center>

\begin{align*}
\mathbf{h}_t &= f(\mathbf{h}_{t-1}, \mathbf{x}_t) \quad \text{for } t = 1, 2, ..., T \\
\mathbf{c} &= \mathbf{h}_T, (\mathbf{c}_T) \quad \text{(context vector)}
\end{align*}

<center> Decocer </center>

\begin{align*}
\mathbf{s}_t &= f(\mathbf{s}_{t-1}, \mathbf{y}_{t-1}, \mathbf{c}) \quad \text{for } t = 1, 2, ..., T' \\
\hat{\mathbf{y}}_t &= \mathrm{softmax}(\mathbf{W} \cdot \mathbf{s}_t + \mathbf{b})
\end{align*}

h_t-1: encoder의 이전 hidden state <br>
h_t: encoder의 현재 hidden state (단어의 의미와 문맥이 반영된 벡터) <br>
f: RNN/LSTM/GRU 레이어  <br>
c: h_t(RNN) / h_t, c_t (LSTM) <br>

s_t-1: decoder의 이전 hidden state  <br>
y_t-1: decoder가 직전 시점에서 생성한 단어 또는 정답 단어  <br>
s_t: decoder의 현재 hidden state <br>
y_t: decoder가 현재 생성한 단어 또는 정답 (전체 단어 사전 중에서 각 단어가 현재 나올 확률 분포)

### 🎹 작곡가, 작사가 비유를 통한 이해

#### 🔮 이야기를 가사로 만드는 상황

요소 | 실제 의미 | 비유
---|---|---
Encoder | 입력 시퀀스를 문맥을 담은 벡터로 압축하는 모델 | 작곡가가 이야기를 보고 멜로디를 만든다
Decoder | Encoder로 부터 얻은 context vector를 바탕으로 출력 시퀀스 생성| 작사가가 멜로디에 맞춰 가사를 하나씩 쓴다
Context vector | Encoder가 만든 문장의 의미 요약 정보 (hidden + cell) | 멜로디와 음악의 전체 분위기 및 정서 등 음악에 대한 정보
Encoder Hidden State | 마지막 시점의 출력 (현재 표현) | 지금 재생되고 있는 멜로디
Encoder Cell State | RNN 계열 모델의 내부 장기 기억 상태 | 이야기의 전체의 흐름, 주제, 정서 같은 전체적인 정보
SOS | 시퀀스 시작 신호 (디코더 첫 입력) | 작곡가가 작사가에게 '자 이제 만들어보자'라고 말하는 것
EOS | 시퀀스 종료 신호 (생성 완료 시점) | 작곡가가 작사가에게 '여기까지 잘 완성했어'라고 말하는 것
이전 단어 | 디코더의 입력으로 사용되는 직전 단어 | 멜로디를 듣고 직전에 쓴 가사 단어 (ex: 'you are my ~')
정답 단어 | 현재 시점에서 예측해야 할 다음 단어 | 지금 멜로디에 어울리는 가사 (ex: 'sunshine')
정답 비교 | 예측 결과와 정답의 비교 (loss 계산) | 작사가가 만든 가사가 잘 만들고 있는지 정답 가사를 보고 비교하기
Teacher Forcing | 예측 대신 정답 단어를 다음 입력으로 사용 | 정답 가사 확인하기

````python
import random

# 숫자 → 영어 단어 사전
num2word = {
    '0': 'zero', '1': 'one', '2': 'two', '3': 'three', '4': 'four',
    '5': 'five', '6': 'six', '7': 'seven', '8': 'eight', '9': 'nine'
}

def generate_data(n=1000, max_len=5):
    data = []
    for _ in range(n):
        length = random.randint(1, max_len)
        digits = [str(random.randint(0, 9)) for _ in range(length)]
        input_seq = ' '.join(digits)
        output_seq = ' '.join([num2word[d] for d in digits])
        data.append((input_seq, output_seq))
    return data

train_data = generate_data(n=5000)
val_data = generate_data(n=500)
````

#### 2) 단어 사전 만들기

````python
def build_vocab(sequence_list, min_freq = 2, mode='target'):
    from collections import Counter
    counter = Counter()

    start_idx = 4 if mode == 'target' else 2;

    for seq in sequence_list:
        counter.update(seq)

    vocab = {word: idx for idx, (word, count) in enumerate(counter.items(), start=start_idx) if count > min_freq}
    vocab['<PAD>'] = 0
    vocab['<UNK>'] = 1

    if mode == 'target':
        vocab['<SOS>'] = 2  # Start of sequence
        vocab['<EOS>'] = 3  # End of sequence

    return vocab
````

#### 3) 정수 인코딩 함수 만들기

````python
def encode(tokens, vocab, max_len, mode='target'):
    if mode == 'target':
        labeled_token = ['<SOS>'] + tokens[:max_len] + ['<EOS>']
        labeled_token += ['<PAD>'] * (max_len - len(tokens))
    elif mode == 'input':
        labeled_token = tokens[:max_len]
        labeled_token += ['<PAD>'] * (max_len - len(tokens))
    else:
        print("Mode is Not Valid")

    return [vocab.get(token, vocab["<PAD>"]) for token in labeled_token]
````

#### 4) Encoder 구현하기

### Encoder

> 입력 시퀀스를 전체로 받아서 하나의 벡터로 압축함 (정보를 요약) <br>
> 입력된 정보를 하나의 덩어리로 인코딩하는 역할

````python
import torch.nn as nn
class Encodcer(nn.Module):
    """
        Encoder는 Context vector를 출력하는 것이 목표 (hidden state, cell state)
    """
    def __init__(self, vocab_size, embedding_dim, hidden_dim, num_layer, dropout=0.5):
        self.embedding = nn.Embedding(vocab_size, embedding_dim)    # (Batch, sequence, embedding_dim)
        self.lstm = nn.LSTM(input_size= embedding_dim,
                            hidden_size= hidden_dim,
                            num_layers= num_layer,
                            dropout=dropout if num_layer > 1 else 0.0,
                            batch_first=True) # (Batch, sequence, embed_dim) @ (embed_dim, hidden) -> (Batch, sequence, hidden)

    def forward(self, input_sequence):
        embed = self.embedding(input_sequence)
        output, (hidden, cell) = self.lstm(embed)

        return output, (hidden, cell)
````

#### 5) Decoder 구현하기

### Decoder

> 압축된 정보를 내가 원하는 형태의 정보로 풀어내는 역할 <br>
> 이전에 생성된 토큰을 받아 다음 토큰을 하나씩 생성하는 구조 <br>
> 자기 자신의 이전 출력을 다음 입력으로 사용하는 방식으로 Auto-regressive라고 함

✅ Decoder는 예측 모델이 아니라 생성 모델!

````python
import torch.nn as nn

class Decodcer(nn.Module):
    """
        Encoder는 Context vector를 출력하는 것이 목표 (hidden state, cell state)
    """
    def __init__(self, vocab_size, embedding_dim, hidden_dim, num_layer, dropout=0.5):
        self.embedding = nn.Embedding(vocab_size, embedding_dim)    # (Batch, sequence, embedding_dim)
        self.lstm = nn.LSTM(input_size= embedding_dim,
                            hidden_size= hidden_dim,
                            num_layers= num_layer,
                            dropout=dropout if num_layer > 1 else 0.0,
                            batch_first=True) # (Batch, sequence, embed_dim) @ (embed_dim, hidden) -> (Batch, sequence, hidden)
        self.fc = nn.Linear(hidden_dim, vocab_size) # 최종적으로 단어사전 크기의 확률 리스트로 출력 -> argmax로 출력

    def forward(self, decoder_token, hidden, cell):     # decoder_token: 정답일수도 있고, 예측값일수도 있는 값 -> (Batch, 1) 로 강제 변환 필요
        decoder_token = decoder_token.unsqueeze(1)
        embed = self.embedding(decoder_token)           # 예측한 하나의 sequence - (Batch, 1, embed_dim)
        output, (hidden, cell) = self.lstm(embed)       # output -> (Batch, 1, hidden)

        prediction = self.fc(output.squeeze(1))

        return output, (hidden, cell)
````

````python
class EarlyStopping:
    def __init__(self, patience=5, delta=0.0, checkpoint_path='checkpoint.pt'):
        self.patience = patience
        self.delta = delta
        self.checkpoint_path = checkpoint_path

        self.best_loss = float('inf')
        self.counter = 0
        self.early_stop = False

    def __call__(self, val_loss, model):
        if val_loss < self.best_loss - self.delta:
            self.best_loss = val_loss
            self.counter = 0
            self._save_checkpoint(model)
        else:
            self.counter += 1
            print(f"EarlyStopping counter: {self.counter} / {self.patience}")
            if self.counter >= self.patience:
                self.early_stop = True
        return self.early_stop

    def _save_checkpoint(self, model):
        torch.save(model.state_dict(), self.checkpoint_path)
````

### BLEU (Bilingual Evaluation Understudy)

> 기계 번역 결과와 사람이 직접 번역한 결과가 얼마나 유사한지 비교하여 번역에 대한 성능을 측정하는 방법 <br>
> n-gram 단위로 모델의 예측 결과와 정답이 얼마나 겹치는지 측정 <br>
> precision 중심 평가이기 때문에 예측에서 정답이 얼마나 많이 나왔는가를 측정

참고: https://aclanthology.org/P02-1040.pdf

$$
P_n = \frac{ \sum_{w_n \in \text{Candidate}} \min(\text{Count}_{\text{Cand}}(w_n), \text{Count}_{\text{Ref}}(w_n)) }{ \sum_{w_n \in \text{Candidate}} \text{Count}_{\text{Cand}}(w_n) }
$$


<center>
Wn: 길이가 n인 n-gram
</center>

<center>
CountCand(Wn): 예측 문장에서 n-gram의 각 단어별 등장 횟수
</center>

<center>
CountRef(Wn): 참조 문장에서 n-gram의 각 단어별 등장 횟수
</center>

<center>
Pn: n-gram일때 Precision
</center>

$$
BP =
\begin{cases}
1 & \text{if } c > r \\\\
\exp\left(1 - \frac{r}{c}\right) & \text{if } c \le r
\end{cases}
$$

<center>
c: 예측 문장의 길이
r: 참조 문장의 길이
</center>

$$
\text{BLEU} = BP \times \exp\left( \sum_{n=1}^{N} w_n \cdot \log P_n \right)
$$

$$
w_n = \frac{1}{N}
$$

예시)

Reference: 나는 밥을 먹는다 나는 배가 부르다 <br>
Candidate: 나는 밥을 먹는다 너는 배가 아프다

Precision(1-gram)

단어 | 예측에 나온 횟수 | 참조에 나온 횟수 | Clipped 횟수
---|---|---|---
나는 | 1  | 2 | 1  
밥을 | 1  | 1 | 1
먹는다 | 1 | 1 | 1
너는 | 1 | 0 | 0
배가 | 1 | 1 | 1
아프다 | 1 | 0 | 0

$$
P_1 = \frac{ \min(1,2) + \min(1,1) + \min(1,1) + \min(1,0) + \min(1,1) + \min(1,0)}{6} = \frac{4}{6}
$$

Precision(2-gram)

단어 | 예측에 나온 횟수 | 참조에 나온 횟수 | Clipped 횟수
---|---|---|---
나는 밥을| 1  | 1 | 1  
밥을 먹는다| 1  | 1 | 1
먹는다 너는| 1 | 0 | 0
너는 배가| 1 | 0 | 0
배가 아프다| 1 | 0 | 0

$$
P_2 = \frac{ \min(1,1) + \min(1,1) + \min(1,0) + \min(1,0) + \min(1,0) }{5} = \frac{2}{5}
$$

$$
BP = \exp\left(1 - \frac{6}{6}\right) = \exp(0) = 1
$$


$$
P_1 = \frac{4}{6} = 0.6667,\quad
P_2 = \frac{2}{5} = 0.4,\quad
BP = 1
$$

$$
\text{BLEU-2} = BP \cdot \exp\left( \frac{1}{2} \log P_1 + \frac{1}{2} \log P_2 \right)
= \exp(-0.6609) \approx \boxed{0.5167}
$$


### ROUGE (Recall-Oriented Understudy for Gisting Evaluation)

> BLEU와 반대로 n-gram 기반의 recall을 측정하는 지표 <br>
> 중요한 정보들이 포함됐는지에 집중하기 때문에 1, 2, .. n-gram 모두를 활용한 지표 계산은 하지 않는다 <br>
> 따라서 기계 번역의 성능 지표로 사용할 수 있긴 하지만 문자 요약 성능 지표로 많이 사용함

### ROUNGE-1

$$
P_1 = \frac{ \sum_{w_n \in \text{Candidate}} \min(\text{Count}_{\text{Cand}}(w_n), \text{Count}_{\text{Ref}}(w_n)) }{ \sum_{w_n \in \text{Reference}} \text{Count}_{\text{Ref}}(w_n) }
$$


Reference: 나는 밥을 먹는다 나는 배가 부르다 <br>
Candidate: 나는 밥을 먹는다 너는 배가 아프다

$$
P_1 = \frac{ \min(2,1) + \min(1,1) + \min(1,1) + \min(1,1) + \min(1,0) }{6} = \frac{4}{6}
