```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt     # pyplot 붙여줘야함
import seaborn as sns
```

```python
# 1. 데이터 로드

file_path = "C:/Users/human/ku-python/data/titanic.csv"
df = pd.read_csv(file_path)

df
```

```python
#2. 데이터 요약

print("데이터 요약: ")
print(df.describe(include='all'))
print("\n\n데이터 정보: ")
df.info()   # 데이터 타입과 결측값 확인
```

```python
# 3. 생존자 / 사망자 수 비교

plt.figure(figsize=(8, 5))
sns.countplot(x="Survived", data=df, palette="coolwarm")
plt.xlabel("Survival Status (0 = Deceased, 1 = Survived)")
plt.ylabel("Count")
plt.title("Survive / Deceased Count")
plt.xticks(ticks=[0, 1], labels=["Deceased", "Survived"])

plt.show()
```

```python
# 4. 성별 생존율 비교

plt.figure(figsize=(8, 5))
sns.barplot(x="Sex", y="Survived", data=df, palette="pastel")
plt.xlabel("Sex")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Gender")

plt.show()
```

```python
# Pclass distinct(1, 2, 3)
# Pclass의 특정값들 조회

distinct_pclass = df['Pclass']
distinct_pclass.unique()

# 또는
distinct_pclass = df['Pclass'].unique()
print(distinct_pclass)
```

```python
# 5. 좌석 등급(Pclass)별 생존율 비교 distinct(1, 2, 3)

plt.figure(figsize=(8, 5))
sns.barplot(x="Pclass", y="Survived", data=df, palette="muted")

plt.xlabel("Pclass")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Passenger Class")

plt.xticks(ticks=[0, 1, 2], labels=["First Class", "Second Class", "Third Class"])

plt.show()
```

```python
# Age의 unique값 확인

distinct_age = df["Age"].unique()
distinct_age
```

```python
# 6. 연령별 생존율 비교

plt.figure(figsize=(10, 6))
sns.histplot(df[df["Survived"] == 1]["Age"].dropna(), bins=30, kde=True, color="blue", label="Survived", alpha=0.6)
sns.histplot(df[df["Survived"] == 0]["Age"].dropna(), bins=30, kde=True, color="red", label="Deceased", alpha=0.6)

plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Age Distribution: Survived / Deceased")
plt.legend()

plt.show()
```

```python
# 운임(Fare)의 unique값 조회

distinct_fare = df["Fare"].unique()
distinct_fare
```

```python
# 7. 운임(Fare)의 분포 비교

plt.figure(figsize=(10, 6))
sns.boxplot(x="Survived", y="Fare", data=df, palette="coolwarm")
plt.xlabel("Servival Status (0 = Deceased, 1 = Servived)")
plt.ylabel("Fare")
plt.title("Fare Distribution by Survival Status")
plt.xticks(ticks=[0, 1], labels=["Deceased", "Survived"])

plt.show()
```

```python
# 8. Pclass(승객 등급)과 성별을 동시에 고려한 생존율 비교

plt.figure(figsize=(8, 5))
sns.catplot(x="Pclass", y="Survived", hue="Sex", data=df, kind="bar", palette="viridis")
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Passenger Class and Gender")


plt.show()
```

```python
# 특정값 조회 | Embarked : 탑승한 항구``
distinct_Embarked = df["Embarked"].unique()
distinct_Embarked
```

```python
# 9. 출발 항구(Embarked) 별 생존율 비교

plt.figure(figsize=(8, 5))
sns.barplot(x="Embarked", y="Survived", data=df, palette="Blues")
plt.xlabel("Embarkation Port")
plt.ylabel("Servival Rate")

plt.show()
```

```python
# SibSp: 타이타닉 호에 동승한 자매(sibling)/ 배우자(spouse)의 수
# Parch: 타이타닉 호에 동승한 부모 / 자식의 수
# 즉 SibSp + Parch == 0이면 가족 동반 없이 혼자 탐 / 1이면 가족과 동행함

# 10. 가족동반 여부와 생존율 비교
df["Family"] = df["SibSp"] + df["Parch"]

plt.Figure(figsize=(8, 5))
sns.barplot(x="Family", y="Survived", data=df, palette="coolwarm")
plt.xlabel("Number of Family Member Onboard")
plt.ylabel("Survival Rate")
plt.title("")

plt.show()
```
