```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt     # pyplot 붙여줘야함
import seaborn as sns
```

```python
# 샘플 데이터 생성

data = {
    "카테고리": ["A", "B", "C", "D"],
    "값": [100, 200, 150, 250]
}

df = pd.DataFrame(data)

# print(df)

# 기본 막대 그래프 생성

# 한글 깨짐 방지용 Font 지정 | 맑음 고딕 가져오기 | 처음 한번만 지정해 주면 됨
plt.rcParams['font.family'] = 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False

plt.bar(df["카테고리"], df["값"], color=['blue', 'green', 'red', 'purple'])
plt.xlabel("카테고리")
plt.ylabel("값")
plt.title("기본 막대 그래프")

plt.show()
```

```python
x = np.linspace(0, 10, 100)
y = np.sin(x)
colors = np.random.rand(50)

plt.figure(figsize=(8, 5))
plt.plot(x, y, label="Sine Wave", color='b')
plt.xlabel("X 축")
plt.ylabel("Y 축")
plt.title("선 그래프 예제")
plt.legend()

plt.show()
```

```python
# 산점도 그래프 예제

np.random.seed(42)
x = np.random.rand(50)
y = np.random.rand(50)

colors = np.random.rand(50)


plt.scatter(x, y, c=colors, cmap='viridis', alpha=0.7, edgecolors='black')
plt.xlabel("X 값")
plt.ylabel("Y 값")
plt.title("선점도 예제")
plt.colorbar(label="색상 값")
plt.show()
```
