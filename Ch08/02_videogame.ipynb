```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
vg = pd.read_csv("C:\\Users\\human\\ku-python\\data\\vgsales.csv")
df = pd.DataFrame(vg)
df_nonZero = df.drop(columns='Year').dropna()
df_nonZero.isna().sum()
```

```python
df_nonZero.head()
```

```python
df_nonZero['Genre'].unique()
```

```python
plt.Figure(figsize=(12, 8))
sns.barplot(x='Genre', y='Global_Sales', data=df_nonZero, palette='Set1')

plt.legend()
plt.show()
```

```python
df_nonZero['Platform'].unique()
```

```python
dict = {}
for i, p in enumerate(df_nonZero['Platform'].unique()):
    dict[p] = i
df_nonZero['Platform'] = df_nonZero['Platform'].map(dict)
df_nonZero
```

```python
dict = {}
for i, p in enumerate(df_nonZero['Publisher'].unique()):
    dict[p] = i
print(dict)
```

```python
df_nonZero['Publisher'] = df_nonZero['Publisher'].map(dict)
df_nonZero
```

```python
df_nonZero['Genre'] = df_nonZero['Genre'].map({'Sports': 1, 'Platform': 2, 'Racing': 3, 'Role-Playing': 4,
                                               'Puzzle': 5, 'Misc': 6, 'Shooter': 7, 'Simulation': 8, 'Action': 9
                                               , 'Fighting': 10, 'Adventure': 11, 'Strategy': 12})
df_nonZero
```

```python
plt.Figure(figsize=(12, 8))
sns.barplot(x='Genre', y='Global_Sales', data=df_nonZero, palette='Set1')

plt.show()
```

```python
vg_numberic = df_nonZero.select_dtypes('number')
correlation_matrix = vg_numberic.corr()

plt.Figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
```
