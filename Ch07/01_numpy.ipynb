```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import ndimage
```

## NUMPY 기초

```python
# 리스트를 Numpy 배열로 변환
list = [1, 2, 3, 4, 5]
arr1 = np.array(list)


# 2차원 배열을 생성
arr1D1 = np.array([1, 2, 3])
arr1D2 = np.array([4, 5, 6])

arr2D1 = np.array([arr1D1, arr1D2])
arr2D2 = np.array([[11, 21, 31], [11, 12, 13]])


# 0으로 채워진 배열 생성
zeros = np.zeros((3, 3))


# 1로 채워진 배열 생성
ones = np.ones((2, 4))


# 특정 값으로 채운 배열 생성
full = np.full((2, 3), 7)


print(full)
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

# 기본 연산
print(arr1 + arr2)  # 요소별 덧셈
print(arr1 - arr2)  # 요소별 뺄셈
print(arr1 * arr2)  # 요소별 곱셈
print(arr1 / arr2)  # 요소별 나눗셈


# 스칼라 연산
print(arr1 * 3)
print(arr1 + 5)
```

```python
arr = np.array([[10, 20, 30], [40, 50, 60]])
print(arr)

# 특정 원소 접근
print(arr[0, 1])    # 20
print(arr[0][1])    # 기존 방식도 가능


# 슬라이싱
sliceArr1 = arr[:, 1]
print(sliceArr1)    # 각 차원의 인덱스 1번을 꺼냄

sliceArr2 = arr[1, :]
print(sliceArr2)    # 40, 50, 60
```

```python
# 2차원 배열
arr = np.array([[1, 2, 3], [4, 5, 6]])

# 1차원 배열
vec = np.array([10, 20, 30])

result = arr + vec  # 1차원 배열이 2차원 배열의 내부 배열에 각각 한번씩 더해짐
                    # arr[0] + vec , arr[1] + vec
print(result)
```

```python
# 유니버셜 함수

arr = np.array([1, 4, 9, 16])
print(np.sqrt(arr)) # 제곱근
print(np.exp(arr))  # 지수함수
print(np.log(arr))  # 로그함수
```

```python
arr = np.array([3, 1, 4, 1, 5, 9])
print(np.sort(arr))     # 정렬 (값)
print(np.argsort(arr))  # 정렬 후의 값들의 정렬전 인덱스를 반환
        # 정렬전    3| 1| 4| 1| 5| 9
        # 인덱스    0| 1| 2| 3| 4| 5
    
        # 정렬후    1| 1| 3| 4| 5| 9
        # 출력값    1| 3| 0| 2| 4| 5
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

print(np.dot(A, B))     # 행렬 곱
print(np.linalg.inv(A)) # 역행렬
print(np.linalg.det(A)) # 행렬식
```

```python
# 대용량 데이터 처리

large_data = np.random.rand(10000000) # 랜덤값 10000000개 생성 - 대용량 데이터
print(large_data.mean())    # 평균계산
print(large_data.std())     # 표준편차
```

## NUMPY 실무 예제

```python
# 흑백 이미지 데이터 생성
image = np.random.rand(100, 100)
plt.imshow(image, cmap='gray')
plt.show()


# 이미지 회전
rotated_image = ndimage.rotate(image, 45)
plt.imshow(rotated_image, cmap='gray')
plt.show()
```

```python
stock_prices = np.array([100, 102, 98, 105, 110])   # 주가 데이터(예: 5일간 증가)

# 변동률(%) 계산: (다음날 가격 - 현재가격) / 현재가격 * 100
# stock_prices[1:] : [102, 98, 105, 110] (두번째 요소부터 끝까지)
# stock_prices[:-1] : [100, 102, 98, 105] (첫번째 요소부터 끝-1 까지)

result = (stock_prices[1:] - stock_prices[:-1]) / stock_prices[:-1] * 100
print(result)
```
