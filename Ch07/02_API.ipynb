```python
# import pandas as pd
# import numpy as np
# import matplotlib.pyplot as plt
# from scipy import ndimage
import requests
```

```python
# GET 요청: JSONPlaceholder의 posts 엔드포인트에서 데이터 가져오기
url = "https://jsonplaceholder.typicode.com/posts/1"
# url = "http://corp.megastudy.net/2234234"
res = requests.get(url)

# 응답 확인
if res.status_code == 200:
    print("GET 요청 성공!")
    print("응답 데이터 :", res.json())
else:
    print("GET 요청 실패", res.status_code)


# POST 
post_url = "https://jsonplaceholder.typicode.com/posts"
data = {
    "title": "파이썬 API 요청 from 구완모",
    "body": "request 라이브러리를 사용한 POST 요청 예제입니다.",
    "userID": 1
}

post_res = requests.post(post_url, json=data)

# 응답 확인
if post_res.status_code == 201:
    print("POST 요청 성공!")
    print("응답 데이터 :", post_res.json())
else:
    print("POST 요청 실패", post_res.status_code)
```
