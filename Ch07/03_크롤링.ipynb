```python
import pandas as pd
import requests as req
from bs4 import BeautifulSoup
from datetime import datetime
```

```python
data = []

for page in range(1, 43):
    # YES24 베스트셀러 페이지 URL(1페이지)
    url = f"https://www.yes24.com/Product/Category/BestSeller?categoryNumber=001&pageNumber={page}&pageSize=24"

    # 웹 요청시 차단 방비 agent 추가
    headers = {"User_Agent": "Mozilla/5.0"}
    res = req.get(url, headers=headers)

    # 응답 확인
    if res.status_code == 200:
        soup = BeautifulSoup(res.text, "html.parser")

        book_titles = soup.select(".gd_name")
        book_prices = soup.select("strong > em")
        book_authors = soup.select("span.authPub.info_auth > a:nth-child(1)")
        book_publisher = soup.select("span.authPub.info_pub > a")
        book_sells = soup.select("span.saleNum")
        book_reviews = soup.select("span.rating_rvCount > a > em.txC_blue")
        book_stars = soup.select("span.rating_grade > em")

        books = zip(book_titles, book_prices, book_authors, book_publisher, book_sells, book_reviews, book_stars)

        for idx, (title, price, author, publisher, sell, review, star) in enumerate(books, start=1):        

            data.append({"title" : title.get_text(), "price": price.get_text(), "author": author.get_text(),
                        "publisher": publisher.get_text(), "sell": sell.get_text().split("수 ")[1].strip(),
                        "review": review.get_text(), "star": star.get_text()})

        print(f"{page} is Done! --- {datetime.now().strftime("%Y-%M-%D %H:%M:%S")}")

    else:
        print(f"요청 실패: {res.status_code}")

df = pd.DataFrame(data)
df
```

```python
data = []

for page in range(1, 283):

    URL = f"https://search.kyobobook.co.kr/search?keyword=%EB%A8%B8%EC%8B%A0%EB%9F%AC%EB%8B%9D&target=total&gbCode=TOT&page={page}"

    headers = {"User_Agent": "Mozilla/5.0"}
    res = req.get(URL, headers=headers)
    
    if res.status_code == 200:
        soup = BeautifulSoup(res.text, "html.parser")
        
        book_titles = soup.select(".prod_info > span:nth-child(2)")
        book_prices = soup.select(".price > .val")
    
        for idx, (title, price) in enumerate(
                zip(book_titles, book_prices), start=1
            ):
                data.append({"title": title.get_text(strip=True), "price": price.get_text(strip=True)})
        print(f"{page}페이지 처리 완료!! --- {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")
    else:
            print(f"응답 실패: {res.status_code}")    

df = pd.DataFrame(data)
```
