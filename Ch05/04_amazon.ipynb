```python
data.replace({"|": 0})
data['rating'] = data['rating'].astype('float32')
```

```python
data['rating_count'] = data['rating_count'].fillna('0')
```

```python
data['rating_count'] = data['rating_count'].str.replace(",", "").astype('int32')
```

```python
filtered_data = data[data['rating'] >= 4.5]
```
