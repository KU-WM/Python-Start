```python
# null값 합산으로 nan값 찾기
data1.isnull().sum()
```

```python
# nan값 없이 출력
data1.dropna()

# nan값을 없애고 기존 데이터에 덮어씌우기
data1.dropna(inplace=True)

# nan값 처리 방식 / any - 값들중 nan이 하나라도 있으면 삭제, all - 모든값이 nan인 경우 삭제
data1.dropna(how='all')
```
