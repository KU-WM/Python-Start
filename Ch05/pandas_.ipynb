### 데이터 분석 전 확인하는 5가지

1. info : 데이터의 기본 정보를 보여준다
2. describe : 숫자형태인 데이터에 대해서 기본 통계값을 보여준다
3. head : default로 앞에서 5개 데이터만 불러온다 (앞에서부터 보고싶은 데이터 갯수 입력 가능)
4. tail : head와 반대로 뒤에서 부터 데이터를 불러온다
5. sample : 랜덤으로 하나의 데이터를 불러온다

> 데이터를 분석하기 전에 분석할 가치가 있는 데이터인지 확인하기 위한 작업이다 <br>
> 1. 데이터 양이 충분한가? -> 데이터 수집이 더 필요할 수 있기 때문 
> 2. 결측 데이터가 너무 많지는 않은가? -> 데이터 활용 가치가 없을 수 있기 때문 
> 3. 데이터의 크기는 얼만한가? -> 데이터 크기가 너무 크면 오류가 나거나 실행 속도가 느려질 수 있기 때문

### Tidy data
> 깔끔하게 정제된 데이터. 이상적인 형식의 데이터 구조. <br>
> 수집한 데이터가 충분히 정제 되어 Tidy data로 만들어져야, 분석과 시각화가 용이해진다

[Tidy data의 조건]
1. 각 변 수는 하나의 열에 있어야 한다 
2. 각 관측값은 하나의 행에 있어야 한다 
3. 각 관측 단위는 하나의 테이블로 구성되어야 한다

https://vita.had.co.nz/papers/tidy-data.pdf

### Pivot_table 
> groupby 연산과 유사하지만 <br>
> 좀 더 복잡한 다차원 데이터를 그룹화 하는데 가독성이 더 좋음 <br>
> 결측치 처리가 옵션으로 사용 할 수 있어서 좀 더 편리


1. index → 그룹화할 기준 컬럼 선정 
2. columns → 새로운 열로 변환할 컬럼 
3. values → 집계할 값 (수치형 데이터)
4. aggfunc → 집계 함수 

```python
data = pd.read_csv('c:/Users/human/ku-python/data/treatment.csv')
```

```python
data.head()
```

```python
data.pivot_table(index='name', values=['treatmenta', 'treatmentb'], aggfunc='mean') #  data.groupby('name').mean()와 같은 연산
```

```python
data.pivot_table(index=['name'], values=['treatmenta','treatmentb'], aggfunc='mean', fill_value=50) 
```

### Crosstab
> 빈도 분석에 특화된 함수

1. index → 행 기준 (Y축)
2. columns → 열 기준 (X축)
3. values → 집계할 수치형 데이터 (기본값은 빈도수)
4. aggfunc → values에 대한 집계 함수 (예: sum, mean 등)
5. margins=True → 총합(Total) 추가
6. normalize=True → 비율(%) 출력

```python
import pandas as pd

data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'C', 'C', 'B'],
    'Year': [2022, 2022, 2023, 2023, 2023, 2022, 2023, 2022],
    'Sales': [100, 150, 200, 130, 170, 90, 120, 160]
}

df = pd.DataFrame(data)
```

```python
pd.crosstab(index=df['Category'], 
            columns=df['Year'], 
            values=df['Sales'], 
            aggfunc='sum', margins=True)
```

```python
pd.crosstab(index=df['Category'], 
            columns=df['Year'], 
            values=df['Sales'], aggfunc='sum', normalize=True)
```

```python
pd.crosstab(index=df['Category'], 
            columns=df['Year'], 
            values=df['Sales'],
            aggfunc='sum', margins=True, normalize=True)
```

```python
df
```

```python
pd.crosstab(df['Category'], df['Year']) # 기본값은 count
```

```python
pd.crosstab(df['Year'], df['Category'])
```

```python
df = pd.read_csv("data/HM_all_stores.csv", low_memory=False)

# 운영 시간 관련 컬럼 추출 (요일별 오픈 시간 데이터)
opening_hours_cols = [col for col in df.columns if '_open_hours' in col]

# 운영 시간 관련 컬럼을 long format으로 변환
df_tidy = df.melt(id_vars=['storeCode', 'storeClass', 'name', 'city', 'country', 'countryCode', 'longitude', 'latitude'],
                  value_vars=opening_hours_cols,
                  var_name='day_of_week',
                  value_name='opening_hours')

# '_open_hours' 접미사를 제거하여 요일 정보만 남기기
df_tidy['day_of_week'] = df_tidy['day_of_week'].str.replace('_open_hours', '')
```

```python
df_tidy.info()
```

```python
df_tidy['day_of_week']=df_tidy['day_of_week'].astype('category')
df_tidy['opening_hours']=df_tidy['opening_hours'].astype('category')
df_tidy['longitude']=df_tidy['longitude'].astype('float32')
df_tidy['latitude']=df_tidy['latitude'].astype('float32')
```

```python
df_tidy.pivot_table(index='country', values='storeCode', aggfunc='count')
```
