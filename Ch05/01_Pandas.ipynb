```python
import numpy as np

# Messy data / 정리되지 않은 데이터
df = pd.DataFrame({
    "Name": ["Alice", "Bob", "Chris", "Tom"],
    "Math": [85, 78, np.nan, 67],
    "Science": [90, 85, 88, 78],
    "English": [88, 92, np.nan, np.nan]
})
```

```python
df
```

```python
# melt = 컬럼을 축약함
# id_vars = 건들이지 않음 || var_name = Column Name을 한 컬럼으로 만듬 || value_name = 실제 데이터들을 한 컬럼으로 만듬

df.melt(id_vars=["Name"], var_name="Subject", value_name="Score")
```

```python
df['Science'] = df['Science'].astype('float32')
```
