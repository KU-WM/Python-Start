1. groupby
    특정 컬럼을 기준으로 그룹화하여 통계 연산을 수행하는 함수
2. pivot table
    엑셀의 피벗 테이블과 유사한 기능으로, 데이터를 그룹화하고 요약하는 함수
3. crosstab
    빈도 분석에 특화된 함수 / 자주 사용하진 않음

```python
import pandas as pd
import numpy as np
```

```python
data1 = pd.read_csv('C:/Users/human/ku-python/data/treatment.csv')
data2 = pd.read_csv("C:/Users/human/ku-python/data/HM_all_stores.csv")
```

```python
# Treatment자료에서 이름별 treatment의 평균계산
data1.groupby('name')['treatmenta'].mean()
```

```python
# 여러 데이터값을 한번에보기
# data.groupby(그룹화할 컬럼)[집계할 컬럼].집계할 컬럼의 연산
data1.groupby('name').agg(['sum', 'mean', 'count', 'min', 'max'])
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'C', 'C', 'B'],
    'Year': [2022, 2022, 2023, 2023, 2023, 2022, 2023, 2022],
    'Sales': [100, 150, 200, 250, 300, 250, 300, 100]
}

data = pd.DataFrame(data)
pd.pivot_table(data, values="Sales", index="Category", columns="Year", aggfunc="sum")
```

```python
data1.pivot_table(columns=['name'], values=['treatmenta', 'treatmentb'], aggfunc='mean', fill_value=0)
```

```python
data = {
    'Category': ['A', 'B', 'A', 'B', 'A', 'C', 'C', 'B'],
    'Year': [2022, 2022, 2023, 2023, 2023, 2022, 2023, 2022],
    'Sales': [100, 150, 200, 250, 300, 250, 300, 100]
}

data = pd.DataFrame(data)

# margin - 현재까지 누적값을 계산해줌
pd.crosstab(index=data['Category'], columns=data['Year'], values=data['Sales'], aggfunc='sum', margins=True)
```

```python
# 기본 설정은 count - 값의 빈도수를 계산해줌
pd.crosstab(data['Category'], data['Year'])
```

```python
# mutation_ID가 12000~12010사이 / patient_id를 기준으로 그룹화 후 score_a 평균

data = pd.read_csv("C:/Users/human/ku-python/data/pathogenicity_scores.csv")
filtered_data = data[(data['Mutation_ID'] >= 12000) & (data['Mutation_ID'] < 12011)]
filtered_data
```

```python
filtered_data.pivot_table(index="Patient_ID", values="SCORE_A", aggfunc="mean")
```
