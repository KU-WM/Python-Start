```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def linear_search(arr, target):
    """선형 탐색으로 target 값을 찾고 탐색 과정을 출력"""
    for index, value in enumerate(arr):
        # print(f"현재 인덱스: {index}, 값: {value}")  # 탐색 과정 출력
        if value == target:
            # print(f"✔ 값 {target}을 찾았습니다! (인덱스: {index})")
            return index  # 찾으면 인덱스 반환
    # print(f"❌ 값 {target}을 찾을 수 없습니다.")
    return -1  # 못 찾으면 -1 반환
```

```python
# 실행 테스트
arr = [1, 33, 17, 29, 3, 50, 12]
target = 50
linear_search(arr, target)
```

```python
def binary_search(arr, target):
    left, right = 0, len(arr) - 1
    step = 1  # 탐색 단계 카운트
    
    while left <= right:
        mid = (left + right) // 2
        # print(f"🔎 Step {step}: 현재 탐색 범위 = {arr[left:right+1]}, 중간값 = {arr[mid]}")
        
        if arr[mid] == target:
            # print(f"🎯 값 {target}을 찾았습니다! (인덱스: {mid})")
            return mid
        elif arr[mid] < target:
            left = mid + 1  # 오른쪽으로 이동
        else:
            right = mid - 1  # 왼쪽으로 이동
        
        step += 1  # 다음 단계로 이동
    
    # print(f"❌ 값 {target}을 찾을 수 없습니다.")
    return -1  # 못 찾으면 -1 반환
```

```python
# 실행 테스트
arr = [1, 2, 3, 4, 6, 7, 8, 9, 10, 11, 13, 14]
target = 3
binary_search(arr, target)
```

```python
# 탐색 과정 시각화 함수
def visualize_search(arr, steps, target, found=True):
    plt.figure(figsize=(8, 5))  # 그래프 크기 설정
    
    for i, (left, mid, right) in enumerate(steps):
        plt.plot(arr, [i] * len(arr), 'ko-', alpha=0.3)  # 전체 리스트 표시
        plt.plot(arr[mid], i, 'ro', markersize=10)  # 현재 중간값 표시
        plt.plot(arr[left:right+1], [i] * (right - left + 1), 'bo-', alpha=0.7)  # 탐색 범위 표시
    
    plt.xlabel("Index")
    plt.ylabel("Iteration Step")
    plt.title(f"Binary Search for {target} {'(Found)' if found else '(Not Found)'}")
    plt.show()
```

```python
# 이진 탐색 함수
def binary_search(arr, target, visualize=False):
    left, right = 0, len(arr) - 1  # 탐색 범위 초기화
    steps = []  # 탐색 과정 저장 리스트
    
    while left <= right:
        mid = (left + right) // 2  # 중간값 계산
        steps.append((left, mid, right))  # 현재 상태 저장
        
        if arr[mid] == target:  # 목표 값과 중간값이 일치하면 반환
            if visualize:
                visualize_search(arr, steps, target)
            return mid
        elif arr[mid] < target:  # 목표 값이 더 크면 오른쪽 탐색
            left = mid + 1
        else:  # 목표 값이 더 작으면 왼쪽 탐색
            right = mid - 1
    
    if visualize:
        visualize_search(arr, steps, target, found=False)
    return -1  # 찾지 못한 경우 -1 반환
```

```python
# 테스트 실행
arr = list(range(1, 21))  # 1부터 20까지 정렬된 리스트
target = 7  # 찾을 값
binary_search(arr, target, visualize=True)  # 이진 탐색 실행 및 시각화 
```

```python
import matplotlib.pyplot as plt
import numpy as np

# 이진 탐색 함수(오버로딩)
def binary_search(arr, target, visualize=False):
    left, right = 0, len(arr) - 1  # 탐색 범위 초기화
    steps = []  # 탐색 과정 저장 리스트

    while left <= right:
        mid = (left + right) // 2  # 중간값 계산
        steps.append((left, mid, right))  # 현재 상태 저장

        if arr[mid] == target:  # 목표 값과 중간값이 일치하면 반환
            if visualize:
                visualize_search(arr, steps, target)
            return mid
        elif arr[mid] < target:  # 목표 값이 더 크면 오른쪽 탐색
            left = mid + 1
        else:  # 목표 값이 더 작으면 왼쪽 탐색
            right = mid - 1

    if visualize:
        visualize_search(arr, steps, target, found=False)
    return -1  # 찾지 못한 경우 -1 반환
```

```python
# 테스트 케이스 설정
test_cases = list(range(1, 11))
linear_times = []
binary_times = []

# 각 테스트 케이스마다 속도 측정
for _ in test_cases:
    data = sorted(random.sample(range(1, 100000), 10000))
    target = data[-1]  # 최악의 경우 (맨 끝 값)

    # 선형 탐색 시간 측정
    start = time.time()
    linear_search(data, target)
    end = time.time()
    linear_times.append(end - start)

    # 이진 탐색 시간 측정
    start = time.time()
    binary_search(data, target)
    end = time.time()
    binary_times.append(end - start)

# 그래프 그리기
plt.figure(figsize=(8, 5))
plt.plot(test_cases, linear_times, 'ro-', label='Linear Search')
plt.plot(test_cases, binary_times, 'bo-', label='Binary Search')

plt.title("Linear Search vs Binary Search Performance")
plt.xlabel("Test Case")
plt.ylabel("Time (seconds)")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()
```
